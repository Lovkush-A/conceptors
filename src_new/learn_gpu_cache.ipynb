```python
import torch

# Simple memory check function
def check_torch_gpu_memory():
    # Total memory available on the GPU
    total_memory = torch.cuda.get_device_properties(0).total_memory / 1e9  # in GB
    
    # Memory reserved by PyTorch's allocator
    reserved_memory = torch.cuda.memory_reserved() / 1e9  # in GB
    
    # Memory actually allocated by PyTorch (subset of reserved)
    allocated_memory = torch.cuda.memory_allocated() / 1e9  # in GB
    
    print(f"Total GPU Memory: {total_memory:.4f} GB")
    print(f"Reserved by PyTorch: {reserved_memory:.4f} GB")
    print(f"Allocated by PyTorch: {allocated_memory:.4f} GB")


import subprocess

def get_gpu_memory():
    result = subprocess.check_output(
        ['nvidia-smi', '--query-gpu=memory.total,memory.used,memory.free', 
         '--format=csv,nounits,noheader'], 
        encoding='utf-8'
    )
    
    # Parse the output
    gpu_memory = [int(x) for x in result.strip().split(',')]
    total, used, free = [mem / 1024 for mem in gpu_memory]  # Convert to GB
    
    print(f"Total GPU memory: {total:.2f} GB")
    print(f"Used GPU memory: {used:.2f} GB")
    print(f"Free GPU memory: {free:.2f} GB")

# Start with clean slate
torch.cuda.empty_cache()

check_torch_gpu_memory()
print("--------------------------------")
get_gpu_memory()
```

```python
# Create a large tensor on GPU
tensor1 = torch.ones(1000, 1000, 1000, device="cuda")
check_torch_gpu_memory()
print("--------------------------------")
get_gpu_memory()
```

```python
# Create another tensor
tensor2 = torch.ones(1000, 1000, 1000, device="cuda")
check_torch_gpu_memory()
print("--------------------------------")
get_gpu_memory()
```

```python
# Delete first tensor
del tensor1
check_torch_gpu_memory()
print("--------------------------------")
get_gpu_memory()
```

```python
# Force garbage collection
torch.cuda.empty_cache()
check_torch_gpu_memory()
print("--------------------------------")
get_gpu_memory()
```
